```python
import numpy as np
import matplotlib.pyplot as plt

t1=np.linspace(-6, -2, 7)
t2=np.linspace(-2, 0, 4)
t3=np.linspace(0, 1, 3)
t4=np.linspace(1, 3, 4)
t5=np.linspace(3, 10, 10)

f1 = 1+((1+t1)**2/5)
f2 = 1+((2+t2)**2/5)
f3 = 1.8+t3*0
f4 = 1+((t4-3)**2/5)
f5 = 1-((t5-3)**2/5)

plt.plot(t1,f1,'o-r', alpha=0.7,label='f1',lw=2)
plt.plot(t2,f2,'+:g',label='f2', lw=2)
plt.plot(t3,f3,'x-m', alpha=0.7,label='f3',lw=2)
plt.plot(t4,f4,'s:b', alpha=0.5,label='f4',lw=2)
plt.plot(t5,f5,'d-y', label='f5',lw=2)

plt.xlabel('t')
plt.ylabel('f')
plt.title('Графики')
plt.grid()
plt.legend()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t=np.linspace(0, 3*(np.pi), 30)
f=((6*(np.e)**(-0.1*t))*(np.sin(t))**5)

t2=np.pi/3
f2=((6*(np.e)**(-0.1*t2))*(np.sin(t2))**5)

t3=7*(np.pi)/6
f3=((6*(np.e)**(-0.1*t3))*(np.sin(t3))**5)

t4=13*(np.pi)/6
f4=((6*(np.e)**(-0.1*t4))*(np.sin(t4))**5)

fmax=np.max(f)
tmax=t[np.argmax(f)]

plt.plot(t,f,lw=2, label='f')
plt.plot(tmax,fmax,'o-g',lw=2,label='Max')

plt.plot(t2,f2,'o-m',lw=2, label='pi/3  7pi/6  13pi/6')
plt.plot(t3,f3,'o-m',lw=2)
plt.plot(t4,f4,'o-m',lw=2)

plt.xlabel('t')
plt.ylabel('f')
plt.title('Графики')
plt.grid()
plt.legend()
```
